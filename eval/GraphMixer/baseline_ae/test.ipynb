```python
import numpy as np

all_labels = np.load("./all_labels.npy", allow_pickle=True)
all_preds = np.load("./all_preds.npy", allow_pickle=True)
```

```python
from sklearn.metrics import RocCurveDisplay, PrecisionRecallDisplay
import matplotlib.pyplot as plt

RocCurveDisplay.from_predictions(all_labels, all_preds)
plt.show()

PrecisionRecallDisplay.from_predictions(all_labels, all_preds)
plt.show()
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, precision_score, recall_score


for t in [0.00005, 0.0005, 0.005, 0.05, 0.5]:

    print("------",t)
    preds = (all_preds > t).astype(int)
    print(confusion_matrix(all_labels, preds))
    print(f1_score(all_labels, preds))
```

```python
losses = all_preds

print(f"Nombre      : {len(losses)}")
print(f"Min         : {losses.min():.6f}")
print(f"Max         : {losses.max():.6f}")
print(f"Moyenne     : {losses.mean():.6f}")
print(f"Médiane     : {np.median(losses):.6f}")
print(f"Écart-type  : {losses.std():.6f}")

for q in [0.01, 0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]:
    print(f"{q:>5.0%} : {np.quantile(losses, q):.6f}")
```
